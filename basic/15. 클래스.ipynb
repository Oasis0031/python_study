```python
# 이름, 나이, 종목
# 플레이어들을 소개하는 메서드

#운동선수 3명을 객체화 하고
#3명을 각각 소개하는 메서드를 사용해서 정보 출력
#ex) OO선수의 나이 OO세 종목 OOO
class Player:
    
    def __init__(self, name, age, sports):
        self.name = name
        self.age = age
        self.sports = sports

    def introduce(self):
        print(f"선수 {self.name}의 나이 {self.age}세, 주 종목은 {self.sports}입니다.")

player1 = Player("이상혁", 30, "LOL")
player2 = Player("김연아", 35, "피겨 스케이트")
player3 = Player("박지성", 45, "축구")

player1.introduce()
player2.introduce()
player3.introduce()
```

```python
#완전 객체 지향형 언어
#Market
#판매

#Customer
#할인율
class Market:
    def __init__(self, market_name, product_name, product_price):
        self.market_name = market_name
        self.product_name = product_name
        self.product_price = product_price

    def sell(self, customer):
        customer.money -= self.product_price

class Customer:
    def __init__(self, name, money, discount):
        self.name = name
        self.money = money
        self.discount = discount
```

```python
cu = Market("CU", "도시락", 10000)
customer = Customer("이규혁", 50000, 10)

cu.sell(customer)
print(customer.money)
```

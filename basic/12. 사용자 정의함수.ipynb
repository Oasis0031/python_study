```python
# 국어, 영어, 수학 점수 평균 구하기
# 자릿수는 2번째 자리부터 반올림
# 사용자의 정보가 입력되면 같이 출력하기
def get_average(**kwards):
    kor, eng, math = kwards.get("kor", 0), kwards.get("eng", 0), kwards.get("math", 0)
    total = kor + eng + math

    average = total / 3
    
    if "round" in kwards:
        average = round(average, kwards["round"])
    
    if "name" in kwards:
        name = kwards["name"]
        return f"{name}님의 평균은: {average}"

    return f"평균은: {average}"
```

```python
user_info = {
    "kor": 100,
    "eng": 80,
    "math": 22,
    "name": "홍길동",
    "round": 5
}
```

```python
get_average(**user_info)
```

```python
# 두 값의 사칙연산 중 계산하고 싶은 값만 리턴해주기

# ex) 10, 20, "-"

# 1. 두 개의 변수는 따로 변수에 받기
# 2. 부호는 따로 keywards 문법으로 받기

nums = {
    "num1": 10,
    "num2": 20
}

def do_math(*args, **kwards):
    calc = kwards.get("calc", "-") 
    if calc == "-":
        result = args[0]-args[1]
        return result
        
print(do_math(nums["num1"], nums["num2"], calc="-"))
```

```python
def cal(num1, num2, **kwards) -> int:

    calc = kwards.get("calc", None)
    calc_set = {"+", "-", "*", "/", "%"}
    is_confirm = calc in calc_set

    if not is_confirm:
        return "부호는 +,-,*,/,%만 사용할 수 있습니다."
    
    num1 = str(num1)
    num2 = str(num2)
    
    if calc:
        return eval(f"{num1}{calc}{num2}")

    return num1, num2

print(cal(10, 20, calc="-"))
```

```python
def get_specific_counts(content, **kwargs):
    # kwargs.values()에는 우리가 찾고 싶은 글자들이 담깁니다.
    # 예: char1="a", char2="p" -> ["a", "p"]
    targets = kwargs.values()
    results = {}
    
    for char in targets:
        # 문자열 내장 메서드인 count()를 활용하면 코드가 더 간결해집니다.
        results[char] = content.count(char)
        
    return results

# 실행 결과 (찾고 싶은 글자를 키워드 인자로 전달)
result = get_specific_counts("apple", first="a", second="p", third="z")
print(result)
# 출력: {'a': 1, 'p': 2, 'z': 0}
```

```python
def count_chars(content):

    result = {}

    for c in content:
        if c not in result:
            result.update({c: content.count(c)})
            
    return result
```

```python
count_chars("apple")
```

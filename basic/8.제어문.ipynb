```python
score = 85  # 예시 점수

grade = ''

if score > 90:
    grade = 'A'
    print("되나")
elif score > 80:
    grade = 'B'
elif score > 70:
    grade = 'C'
else:             # elif 대신 else를 사용합니다.
    grade = 'F'

print(grade)
```

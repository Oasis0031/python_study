## 슬라이싱(slicing)
- 시퀀스 자료구조에서 인덱스 범위부터 인덱스 범위까지 접근하는 문법

```python
data = "이규혁노민균문윤성노규호"
```

```python
#[시작 인덱스:종료 인덱스]
#단, 종료 인덱스를 포함하지 않는다.
data[0:3] 
```

```python
# [:끝까지]
data[-3:]
```

```python
# [처음부터:종료인덱스]
data[:-3]
```

```python
numbers = "123456789"
```

```python
numbers[::2]
```

```python
# 짝수를 가져온다면
# [::스텝]
numbers[1::2]
```

```python
# numbers 뒤집고 싶으면
numbers[::-1]
```

```python
#슬라이싱 실습
#홀수만 가져와서 반대로 뒤집기
numbers[::-2]
```

```python
# 인덱싱과 형변환을 이용해서 -10 만들기
result = "-" + str(int(numbers[0]) + int(numbers[-1]))
print(f"{result}")
```

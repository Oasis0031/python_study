```python
class User:
    name: str
    age:str
    phone:str



    def __init__(self, name, age, phone):
        self.name = name
        self.age = age
        self.phone = phone

    def __str__(self):
        return f"User(name: {self.name}, age: {self.age}, phone: {self.phone})"
```

```python
print(User("홍길동", 20, "010-1234-1234"))
```

```python
# 회사원
class Employee(BaseModel):
    id: int
    name: str
    income: int

class Student(BaseModel):
    id: int
    name: str
    income: int
```

```python
employee_data1 = {
    "id": 1,
    "name" : "홍길동",
    "income" : 100000
}

employee_data2 = {
    "id": 2,
    "name" : "이순신",
    "income" :20000
}

employee1 = Employee(**employee_data1)
employee2 = Employee(**employee_data2)

student_data1 = {
    "id": 1,
    "name" : "김철수",
    "income" : 100000
}

student_data2 = {
    "id": 1,
    "name" : "김영희",
    "income" : 100000
}

student1 = Student(**student_data1)
student2 = Student(**student_data2)
```

```python
print(employee1)
print(employee2)
print(student1)
print(student2)
```

```python
# 완전 객체 지향형 언어
# Market
# 판매
class Market:
    def __init__(self, market_name, product_name, product_price):
        self.market_name = market_name
        self.product_name = product_name
        self.product_price = product_price

    def sell(self, customer):
        customer.money -= self.product_price * (100 - customer.discount) / 100

# Customer
# 할인율
class Customer:
    def __init__(self, name, money, discount):
        self.name = name
        self.money = money
        self.discount = discount
```

```python
cu = Market("CU", "도시락", 10000)
customer = Customer("이규혁", 50000, 10)

cu.sell(customer)
print(customer.money)
```

```python
# Atm, User 객체
# deposit

# Atm.deposit(User, 10000)
# 유저의 돈은 10000 차감,
# 은행에 돈은 10000 입금

# 은행이 가진돈과 유저의 돈 출력
```

```python
class Atm:
    def __init__(self, name):
        self.name = name
        self.users = {}

    def deposit(self, user, money):

        if user.money < money: 
            return "잔액 부족"
        
        if user.name in self.users:
            # 기존의 돈
            self.users.update({user.name: self.users[user.name] + money})
            
        else:
            self.users.update({user.name: money})

        user.money -= money

    def show(self):
        print(self.users)
```

```python
class User:
    def __init__(self, name, money):
        self.name = name
        self.money = money

    def show(self):
        format_money = format(self.money, "_")
        print(f"{self.name}님의 잔 액: {format_money}원")
```

```python
#클래스 메서드
class Korean:
    # 클래스 변수
    country = "한국"

    # 클래스 메서드
    # 인스턴스 X, 클래스 변수 O
    @classmethod # 데코레이터
    def trip(cls, country):
        if cls.country == country:
            print("국내여행")
        else: 
            print("해외여행")          

    # 정적 메서드
    @staticmethod
    def slogan():
        print("Imagin your Korea")
    
```

```python
hong = Korean()

hong.trip("한국")
hong.slogan()

Korean.trip("미국")
Korean.slogan()
```

## 상속(inheritance)

```python
class Person:
    def __init__(self, name):
        self.name = name

class Student(Person):
     def __init__(self, name, age):
         super().__init__(name)
         self.age = age
```

```python
hong = Person("홍길동")
jang = Student("장보고", 20)

# 부모 -> 타입도 같이 전달
# 모든 자식은 부모 타입이다.
# 부모는 자식타입이 아니다.

print(isinstance(hong, Person))
print(isinstance(hong, Student))
print(isinstance(jang, Person))
print(isinstance(jang, Person))
```

```python
class Animal:
    def __init__(self, name, feed, species):
        self.name = name
        self.feed = feed
        self.species = species

class Dog(Animal):
    def __init__(self, name, feed, species):
        super().__init__(name, feed, species)

    def act(self):
        print(f"{self.name}는 산책을 합니다.")

class Cat(Animal):
    def __init__(self, name, feed, species):
        super().__init__(name, feed, species)

    def sleep(self):
        print(f"{self.name}은 잠을 자주 잡니다.")
```

```python
dog = Dog("누렁이", "잡식", "강아지")
dog.act()

cat = Cat("야옹이", "츄르", "고양이")
cat.sleep()
```

```python
class Person:
    def __init__(self, name, age, height, weight):
        self.name = name
        self.age = age
        self.height = height
        self.weight = weight

    def greeting(self):
        print(f"{self.name}(이)가 인사합니다. '반갑습니다.'")

class Student(Person):
    def __init__(self, name, age, height, weight, phone):
        super().__init__(name, age, height, weight)

    def studing(self):
        print(f"{self.name}(이)가 공부합니다.")
    
```

```python
human = Person("인간", 26, 175, 65)
human.greeting()

hakseng = Student("잉여", 17, 150, 40, "010-1234-1234")
hakseng.greeting()
hakseng.studing()
```

## 다형성

```python
class Car:
    def __init__(self, brand, price):
        self.brand = brand
        self.price = price

    def start_engine(self):
        print(f"{self.brand}는 열쇠로 시동을 켭니다.")

    def stop_engine(self):
        print(f"{self.brand}는 열쇠로 시동을 끕니다.")

class SuperCar(Car):
    def start_engine(self):
        print(f"{self.brand}는 음성으로 시동을 켭니다.")

    def stop_engine(self):
        print(f"{self.brand}는 음성으로 시동을 끕니다.")
        
```

```python
matiz = Car("마티즈", 1000000)
matiz.start_engine()
matiz.stop_engine()
```

```python
tesla = SuperCar("테슬라",10000000000)
tesla.start_engine()
tesla.stop_engine()
```

```python
class Phone:
    def __init__(self, brand, price):
        self.brand = brand
        self.price = price

class Phone_2G(Phone):
     def __init__(self, brand, price):
         super().__init__(brand, price)

     def call(self):
         print(f"{self.brand}로 전화를 겁니다.")

class Phone_3G(Phone_2G):
     def __init__(self, brand, price):
         super().__init__(brand, price)

     def internet(self):
         print(f"{self.brand}로 인터넷을 이용합니다.")

class Phone_4G(Phone_3G):
    def __init__(self, brand, price):
         super().__init__(brand, price)

    def video_call(self):
        print(f"{self.brand}로 영상통화를 이용합니다.")
```

```python
# 휴대폰 필드
# 브랜드(brand), 가격(price)
# 2G 전화하기
# 3G 전화하기, 인터넷
# 4G 전화하기, 인터넷, 영상통화

class Phone2G:
    def __init__(self, brand, price):
        self.brand = brand
        self.price = price

    def call(self):
        print(f"{self.brand}은 일반 전화")

class Phone3G(Phone2G):
    def __init__(self, brand, price):
        super().__init__(brand, price)

    def call(self):
        print(f"{self.brand}은 영상 통화도 가능")

    def browse(self):
        print("인터넷 느리게 사용 가능")

class Phone4G(Phone3G):
    def __init__(self, brand, price):
        super().__init__(brand, price)

    def browse(self):
        print("인터넷 사용 가능(속도 개선)")

    def stream_vedio(self):
        print("영상 통화 가능")
```
